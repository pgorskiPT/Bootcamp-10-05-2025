```python
import openpyxl

wb = openpyxl.load_workbook('videogamesales.xlsx')
wb
```

```python
ws = wb['vgsales']
```

```python
ws
```

```python
print("Total number of rows:", ws.max_row)
```

```python
print("total number of columns:", ws.max_column)
```

```python
print("The value in cell A1 is:", ws['A1'])
print("The value in cell A1 is:", ws['A1'].value)  # wartość w komórce
```

```python
values = [ws.cell(row=1, column=i).value for i in range(1, ws.max_column + 1)]  # +1 bo z prawej zbiór otwarty
# excel liczy od 1, python od 0
print(values)
```

```python
data = [ws.cell(row=i, column=2).value for i in range(2, 12)]
print(data)
```
